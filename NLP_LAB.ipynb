patterns in text.

EXP 02 -Implement a basic finite state automaton that recognizes a specific language or pattern. In this
example, we'll create a simple automaton to match strings ending with 'ab' using python.

```python
def fsa_ends_with_ab(string):
    state = 0

    for ch in string:
        if state == 0:
            if ch == 'a':
                state = 1
            else:
                state = 0

        elif state == 1:
            if ch == 'a':
                state = 1
            elif ch == 'b':
                state = 2
            else:
                state = 0

        elif state == 2:
            if ch == 'a':
                state = 1
            elif ch == 'b':
                state = 0
            else:
                state = 0

    if state == 2:
        print("Accepted")
    else:
        print("Rejected")

# Input
s = input("Enter a string: ")
fsa_ends_with_ab(s)
```

simple machine to generate plural forms of English nouns using python.

```python
def pluralize(noun):
    if noun.endswith(('s', 'x', 'z', 'ch', 'sh')):
        return noun + "es"
    elif noun.endswith("y") and noun[-2].lower() not in "aeiou":
        return noun[:-1] + "ies"
    elif noun.endswith("f"):
        return noun[:-1] + "ves"
    elif noun.endswith("fe"):
        return noun[:-2] + "ves"
    else:
        return noun + "s"

word = input("Enter a singular noun: ")
print("Plural form:", pluralize(word))
```

EXP 09 -Implement a rule-based part-of-speech tagging system using regular expressions using python.

```python
import re

def pos_tag(word):
    if re.fullmatch(r"(a|an|the)", word):
        return "DT"      # Determiner
    elif re.fullmatch(r"(is|am|are|was|were)", word):
        return "VB"      # Verb
    elif re.fullmatch(r".*ing", word):
        return "VBG"     # Verb (Gerund)
    elif re.fullmatch(r".*ed", word):
        return "VBD"     # Verb (Past Tense)
    elif re.fullmatch(r".*ly", word):
        return "RB"      # Adverb
    elif re.fullmatch(r".*ous|.*ful|.*able|.*ive", word):
        return "JJ"      # Adjective
    elif re.fullmatch(r".*s", word):
        return "NNS"     # Plural Noun
    else:
        return "NN"      # Noun (Default)

sentence = input("Enter a sentence: ").lower()
words = sentence.split()

print("\nPart-of-Speech Tags:")
for word in words:
    print(word, "->", pos_tag(word))
```

EXP 11 -Implement a simple top-down parser for context-free grammars using python.

```python
class Parser:

    def __init__(self, grammar, start_symbol):
        self.grammar = grammar
        self.start_symbol = start_symbol

    def parse(self, sentence):
        self.sentence = sentence
        self.index = 0
        return self._parse_symbol(self.start_symbol)

    def _parse_symbol(self, symbol):
        if symbol in self.grammar:  # Non-terminal
            for production in self.grammar[symbol]:
                # Store current index for backtracking
                initial_index_for_production = self.index
                if self._parse_production(production):
                    return True
                else:
                    # Backtrack if production fails
                    self.index = initial_index_for_production
            return False
        else:  # Terminal
            if self.index < len(self.sentence) and self.sentence[self.index] == symbol:
                self.index += 1
                return True
            return False

    def _parse_production(self, production):
        initial_index = self.index
        for element in production:
            if not self._parse_symbol(element):
                self.index = initial_index  # Backtrack
                return False
        return True

# Define a simple context-free grammar
grammar = {
    'S': [['A', 'B']],
    'A': [['a']],
    'B': [['b']]
}

# Create a parser instance
parser = Parser(grammar, 'S')

# Test sentences
sentence1 = ['a', 'b']
sentence2 = ['a', 'a', 'b'] # This should fail given the grammar
sentence3 = ['b', 'a'] # This should fail given the grammar

print(f"Parsing {sentence1}: {parser.parse(sentence1)}")
parser = Parser(grammar, 'S') # Reset parser for next sentence
print(f"Parsing {sentence2}: {parser.parse(sentence2)}")
parser = Parser(grammar, 'S') # Reset parser for next sentence
print(f"Parsing {sentence3}: {parser.parse(sentence3)}")
```

EXP 12-Implement an Earley parser for context-free grammars using a simple python program.

```python
class EarleyParser:
    def __init__(self, grammar, start_symbol):
        # Convert all production lists in the grammar to tuples for hashability
        self.grammar = {nt: [tuple(prod) for prod in prods] for nt, prods in grammar.items()}
        self.start_symbol = start_symbol

    def parse(self, words):
        n = len(words)
        chart = [set() for _ in range(n + 1)]

        # Initial state: (S' -> .S, 0, 0). Ensure production is a tuple.
        chart[0].add((('S_prime', (self.start_symbol,)), 0, 0))

        for j in range(n + 1):
            # Iterate until no new states are added to chart[j] in an iteration
            while True:
                size_before_iteration = len(chart[j])
                # Create a list of states to process in this pass. Important to iterate over a copy
                # as the set chart[j] might be modified during iteration.
                states_to_process_in_this_pass = list(chart[j])

                for state in states_to_process_in_this_pass:
                    rule, dot_pos, start_index = state
                    non_terminal, production_tuple = rule

                    # If the dot is at the end, it's a complete state
                    if dot_pos == len(production_tuple):
                        self._complete(chart, state, j)
                    else:
                        next_symbol = production_tuple[dot_pos]
                        if next_symbol in self.grammar:  # Non-terminal
                            self._predict(chart, next_symbol, j)
                        else:  # Terminal
                            # Scan only if there's input left at position j
                            if j < n:
                                self._scan(chart, state, next_symbol, j, words)

                # If no new states were added to chart[j] during this pass, we have reached saturation for this j
                if len(chart[j]) == size_before_iteration:
                    break

        # Check if the start symbol can be parsed across the entire sentence
        # A valid parse exists if chart[n] contains (('S_prime', (S,)), 1, 0)
        final_states = [s for s in chart[n] if s == (('S_prime', (self.start_symbol,)), 1, 0)]
        return len(final_states) > 0

    def _predict(self, chart, non_terminal, j):
        # For each production of the non_terminal, add a new state to chart[j]
        for prod in self.grammar[non_terminal]:
            new_state = ((non_terminal, prod), 0, j)
            if new_state not in chart[j]: # Add only if not already present
                chart[j].add(new_state)

    def _scan(self, chart, state, terminal, j, words):
        # If the next word matches the terminal, advance the dot and add to chart[j+1]
        if j < len(words) and words[j] == terminal:
            rule, dot_pos, start_index = state
            new_state = (rule, dot_pos + 1, start_index)
            if new_state not in chart[j + 1]: # Add only if not already present
                chart[j + 1].add(new_state)

    def _complete(self, chart, completed_state, j):
        # When a rule is completed, look for states that were expecting this non_terminal
        completed_rule, _, completed_start_index = completed_state
        completed_non_terminal = completed_rule[0]

        # Iterate over states in chart[completed_start_index] to find those expecting completed_non_terminal
        for prev_state in list(chart[completed_start_index]): # Iterate over a copy
            prev_rule, prev_dot_pos, prev_start_index = prev_state
            prev_non_terminal, prev_production_tuple = prev_rule

            # If the previous state expects the completed non_terminal at its dot position
            if prev_dot_pos < len(prev_production_tuple) and prev_production_tuple[prev_dot_pos] == completed_non_terminal:
                new_state = (prev_rule, prev_dot_pos + 1, prev_start_index)
                if new_state not in chart[j]: # Add only if not already present
                    chart[j].add(new_state)

# Define a context-free grammar
grammar = {
    'S': [['NP', 'VP']],
    'VP': [['V', 'NP'], ['V']],
    'NP': [['Det', 'N']],
    'V': [['eats']],
    'N': [['cat'], ['fish']],
    'Det': [['the']]
}

# Create an Earley parser instance
parser = EarleyParser(grammar, 'S')

# Test sentences
sentence1 = ['the', 'cat', 'eats', 'the', 'fish']
sentence2 = ['the', 'cat', 'eats']
sentence3 = ['eats', 'fish'] # Should fail
sentence4 = ['the', 'fish', 'eats', 'the', 'cat']

print(f"Parsing {sentence1}: {parser.parse(sentence1)}")
print(f"Parsing {sentence2}: {parser.parse(sentence2)}")
print(f"Parsing {sentence3}: {parser.parse(sentence3)}")
print(f"Parsing {sentence4}: {parser.parse(sentence4)}")
```

EXP 13-Generate a parse tree for a given sentence using a context-free grammar using python program.

```python
class ParseNode:
    """Represents a node in the parse tree."""
    def __init__(self, label, children=None, word=None):
        self.label = label  # The grammar symbol (non-terminal) or terminal word
        self.children = children if children is not None else []
        self.word = word    # The actual terminal word for leaf nodes

    def __repr__(self):
        if self.word:
            return f"('{self.label}', '{self.word}')"
        return f"('{self.label}', {self.children})"

    def pretty_print(self, indent=0):
        """Prints the parse tree in a readable, indented format."""
        prefix = '  ' * indent
        if self.word:
            return f"{prefix}{self.label}: {self.word}"
        else:
            children_str = []
            for child in self.children:
                children_str.append(child.pretty_print(indent + 1))
            return f"{prefix}{self.label}\n" + "\n".join(children_str)

class RecursiveDescentParserWithTree:
    """A simple recursive descent parser that builds a parse tree."""
    def __init__(self, grammar, start_symbol):
        # Grammar: dict from Non-terminal to list of lists (productions)
        # e.g., {'S': [['NP', 'VP']], 'VP': [['V', 'NP'], ['V']]}
        self.grammar = grammar
        self.start_symbol = start_symbol

    def parse(self, sentence):
        self.sentence = sentence
        self.index = 0
        # Attempt to parse the start symbol and get the root node of the tree
        root_node = self._parse_symbol(self.start_symbol)

        # A successful parse means we got a root node AND consumed the entire sentence
        if root_node is not None and self.index == len(self.sentence):
            return root_node
        else:
            return None # Parsing failed or didn't consume entire sentence

    def _parse_symbol(self, symbol):
        """Attempts to parse a given symbol (terminal or non-terminal)."""
        if symbol in self.grammar:  # Non-terminal
            # Try each production rule for this non-terminal
            for production in self.grammar[symbol]:
                initial_index = self.index # Save state for backtracking
                children_nodes = []
                production_success = True

                # Try to parse each element in the production rule
                for element in production:
                    child_node = self._parse_symbol(element)
                    if child_node is not None:
                        children_nodes.append(child_node)
                    else:
                        # If any element fails, this production fails
                        production_success = False
                        self.index = initial_index # Backtrack
                        break

                if production_success:
                    # If the entire production was successfully parsed, create a ParseNode
                    return ParseNode(symbol, children=children_nodes)
            return None # All productions failed for this non-terminal

        else:  # Terminal
            # Check if the current input matches the terminal symbol
            if self.index < len(self.sentence) and self.sentence[self.index] == symbol:
                word_node = ParseNode(symbol, word=self.sentence[self.index])
                self.index += 1
                return word_node
            return None # Terminal mismatch

# Define a context-free grammar
grammar = {
    'S': [['NP', 'VP']],
    'VP': [['V', 'NP'], ['V']],
    'NP': [['Det', 'N']],
    'V': [['eats']],
    'N': [['cat'], ['fish']],
    'Det': [['the']]
}

# Create a parser instance
parser_with_tree = RecursiveDescentParserWithTree(grammar, 'S')

# Test sentences
sentence1 = ['the', 'cat', 'eats', 'the', 'fish']
sentence2 = ['the', 'cat', 'eats']
sentence3 = ['eats', 'fish'] # Should fail
sentence4 = ['the', 'fish', 'eats', 'the', 'cat']

print(f"--- Parsing {sentence1} ---")
tree1 = parser_with_tree.parse(sentence1)
if tree1:
    print("Parse successful. Parse Tree:")
    print(tree1.pretty_print())
else:
    print("Parse failed.")

print(f"\n--- Parsing {sentence2} ---")
parser_with_tree = RecursiveDescentParserWithTree(grammar, 'S') # Reset parser state
tree2 = parser_with_tree.parse(sentence2)
if tree2:
    print("Parse successful. Parse Tree:")
    print(tree2.pretty_print())
else:
    print("Parse failed.")

print(f"\n--- Parsing {sentence3} ---")
parser_with_tree = RecursiveDescentParserWithTree(grammar, 'S') # Reset parser state
tree3 = parser_with_tree.parse(sentence3)
if tree3:
    print("Parse successful. Parse Tree:")
    print(tree3.pretty_print())
else:
    print("Parse failed.")

print(f"\n--- Parsing {sentence4} ---")
parser_with_tree = RecursiveDescentParserWithTree(grammar, 'S') # Reset parser state
tree4 = parser_with_tree.parse(sentence4)
if tree4:
    print("Parse successful. Parse Tree:")
    print(tree4.pretty_print())
else:
    print("Parse failed.")
```

EXP 14-Create a program in python to check for agreement in sentences based on a context-free
grammar's rules.

```python
class ParseNode:
    """Represents a node in the parse tree, now with an 'agreement_feature'."""
    def __init__(self, label, children=None, word=None, agreement_feature=None):
        self.label = label
        self.children = children if children is not None else []
        self.word = word
        self.agreement_feature = agreement_feature # e.g., 'sg', 'pl', 'neutral'

    def __repr__(self):
        if self.word:
            return f"('{self.label}', '{self.word}', num='{self.agreement_feature}')"
        return f"('{self.label}', {self.children}, num='{self.agreement_feature}')"

    def pretty_print(self, indent=0):
        """Prints the parse tree in a readable, indented format."""
        prefix = '  ' * indent
        feature_str = f" (num={self.agreement_feature})" if self.agreement_feature else ""
        if self.word:
            return f"{prefix}{self.label}{feature_str}: {self.word}"
        else:
            children_str = []
            for child in self.children:
                children_str.append(child.pretty_print(indent + 1))
            return f"{prefix}{self.label}{feature_str}\n" + "\n".join(children_str)

class AgreementAwareParser:
    """A recursive descent parser that checks for grammatical agreement."""
    def __init__(self, grammar, start_symbol, terminal_features):
        self.grammar = grammar # dict: Non-terminal -> list of lists (productions)
        self.start_symbol = start_symbol
        # terminal_features: dict mapping word -> {'pos': 'POS_TAG', 'num': 'sg'/'pl'/'neutral'}
        self.terminal_features = terminal_features
        self.sentence = []
        self.index = 0

    def parse(self, sentence):
        """Attempts to parse the entire sentence and returns the root ParseNode if successful."""
        self.sentence = sentence
        self.index = 0
        root_node = self._parse_symbol(self.start_symbol)

        # A successful parse means we got a root node AND consumed the entire sentence
        if root_node is not None and self.index == len(self.sentence):
            return root_node
        else:
            return None # Parsing failed or didn't consume entire sentence

    def _parse_symbol(self, symbol_to_parse):
        """Attempts to parse a given symbol (which can be a non-terminal like 'NP' or a literal word like 'eats')."""

        # Case 1: symbol_to_parse is a Non-terminal (e.g., 'S', 'NP', 'VP', 'V', 'N', 'Det')
        if symbol_to_parse in self.grammar:
            for production in self.grammar[symbol_to_parse]:
                initial_index = self.index # Save state for backtracking if this production fails
                children_nodes = []
                production_success = True

                # Try to parse each element in the production rule (e.g., ['NP', 'VP'] or ['eats'])
                for element_in_production in production:
                    child_node = self._parse_symbol(element_in_production) # Recursive call
                    if child_node is not None:
                        children_nodes.append(child_node)
                    else:
                        # If any element fails, this production fails
                        production_success = False
                        self.index = initial_index # Backtrack to the state before this production started
                        break

                if production_success:
                    # If the entire production was successfully parsed:
                    # For lexical non-terminals (V, N, Det), propagate the agreement feature from their word child.
                    if symbol_to_parse in ['V', 'N', 'Det']:
                        # The child_node for a lexical production (e.g., 'eats') is the actual word's ParseNode.
                        # We use its agreement feature for the parent (V, N, or Det) node.
                        if children_nodes and children_nodes[0].agreement_feature is not None:
                            return ParseNode(symbol_to_parse, children=children_nodes, agreement_feature=children_nodes[0].agreement_feature)
                        else:
                            # Should not happen if terminal_features are well-defined for all words in grammar
                            return ParseNode(symbol_to_parse, children=children_nodes) # No agreement feature
                    else:
                        # For higher-level non-terminals (S, NP, VP), apply specific agreement rules
                        agreement_ok, parent_num_feature = self._check_agreement(symbol_to_parse, children_nodes)
                        if agreement_ok:
                            return ParseNode(symbol_to_parse, children=children_nodes, agreement_feature=parent_num_feature)
                        else:
                            self.index = initial_index # Backtrack if agreement check fails
            return None # All productions for this non-terminal failed

        # Case 2: symbol_to_parse is a Terminal (an actual word, like 'eats', 'the', 'cat')
        # This branch is reached if 'symbol_to_parse' is not a key in self.grammar.
        else:
            if self.index < len(self.sentence):
                current_word_from_sentence = self.sentence[self.index]

                # The key correction: Directly match the literal word from the grammar production
                # with the current word from the input sentence.
                if current_word_from_sentence == symbol_to_parse:
                    # Ensure the word has defined features in terminal_features
                    if current_word_from_sentence in self.terminal_features:
                        features = self.terminal_features[current_word_from_sentence]
                        self.index += 1
                        # The label for this leaf node should be its POS tag (e.g., 'Det', 'N', 'V'),
                        # not the word itself. This is important for parent non-terminals to correctly
                        # identify their children by POS category.
                        return ParseNode(features['pos'], word=current_word_from_sentence, agreement_feature=features.get('num'))
                    else:
                        # Word found in sentence and grammar, but no features defined. This is an error.
                        # For robustness, we could try to parse without features or return None.
                        # Returning None makes the parse fail if features are missing.
                        return None
            return None # Terminal mismatch or end of sentence

    def _check_agreement(self, parent_symbol, children_nodes):
        """Custom logic for checking agreement based on the parent_symbol and its children's features."""
        # This function returns a tuple: (True/False for agreement, determined_number_feature for the parent)

        if parent_symbol == 'S':
            # S -> NP VP: Subject-Verb Agreement. NP's number must agree with VP's number.
            if len(children_nodes) < 2: return False, None # Safety check
            np_node = children_nodes[0] # Assuming NP is the first child
            vp_node = children_nodes[1] # Assuming VP is the second child

            # Both NP and VP must have a number feature for agreement to be checked.
            if np_node.agreement_feature is None or vp_node.agreement_feature is None:
                return False, None # Cannot check agreement if features are missing

            if np_node.agreement_feature == vp_node.agreement_feature:
                return True, None # S doesn't typically carry a number feature itself
            else:
                # Agreement mismatch: e.g., 'cats eats' (pl. NP, sg. VP)
                return False, None

        elif parent_symbol == 'NP':
            # NP -> Det N: Noun Phrase inherits its number from the Noun.
            # Determiner's number must be compatible (neutral or matching N's).
            if len(children_nodes) < 2: return False, None # Safety check
            det_node = children_nodes[0]
            n_node = children_nodes[1]

            if det_node.agreement_feature is None or n_node.agreement_feature is None:
                return False, None # Cannot check agreement if features are missing

            if det_node.agreement_feature == 'neutral' or det_node.agreement_feature == n_node.agreement_feature:
                # NP inherits number from N
                return True, n_node.agreement_feature
            else:
                # Agreement mismatch: e.g., 'a cats' (sg. Det, pl. N)
                return False, None

        elif parent_symbol == 'VP':
            # VP -> V NP or VP -> V: Verb Phrase inherits its number from the Verb.
            if not children_nodes: return False, None # Safety check
            v_node = children_nodes[0]

            if v_node.agreement_feature is None:
                return False, None # Cannot determine VP's number without V's number

            # VP's number is simply the verb's number
            return True, v_node.agreement_feature

        # For other non-terminals or productions, assume agreement is always true and no number feature to propagate
        return True, None


# Define a context-free grammar with POS tags as terminals in the RHS of NP/VP, and actual words in lexical rules.
grammar = {
    'S': [['NP', 'VP']],
    'VP': [['V', 'NP'], ['V']], # 'V' here refers to the non-terminal 'V' which expands to words
    'NP': [['Det', 'N']],        # 'Det', 'N' here refer to the non-terminals 'Det' and 'N'
    'V': [['eats'], ['eat']],    # Lexical rules for verbs
    'N': [['cat'], ['cats'], ['fish']], # Lexical rules for nouns
    'Det': [['the'], ['a']]      # Lexical rules for determiners
}

# Define terminal features: mapping words to their POS and number features
terminal_features = {
    'the': {'pos': 'Det', 'num': 'neutral'}, # 'the' can precede sg or pl nouns
    'a': {'pos': 'Det', 'num': 'sg'},       # 'a' must precede singular nouns
    'cat': {'pos': 'N', 'num': 'sg'},
    'cats': {'pos': 'N', 'num': 'pl'},
    'fish': {'pos': 'N', 'num': 'sg'},      # Treating 'fish' as singular for simplicity in this example
    'eats': {'pos': 'V', 'num': 'sg'},      # Singular verb
    'eat': {'pos': 'V', 'num': 'pl'}        # Plural verb
}

# Create an agreement-aware parser instance
# Note: Each call to parse should ideally be with a fresh parser instance if mutable state is involved
# or reset internal state if parse is to be called multiple times on the same object.

# Test sentences
sentence1 = ['the', 'cat', 'eats', 'the', 'fish'] # Valid: Sg NP, Sg VP
sentence2 = ['the', 'cats', 'eat', 'the', 'fish'] # Valid: Pl NP, Pl VP
sentence3 = ['the', 'cat', 'eat', 'the', 'fish']  # Invalid: Sg NP, Pl VP (agreement fail)
sentence4 = ['the', 'cats', 'eats', 'the', 'fish'] # Invalid: Pl NP, Sg VP (agreement fail)
sentence5 = ['a', 'cat', 'eats'] # Valid: Sg Det, Sg N, Sg V
sentence6 = ['a', 'cats', 'eat'] # Invalid: Sg Det, Pl N (agreement fail in NP)
sentence7 = ['the', 'cat', 'eats'] # Valid: Sg NP, Sg VP (shorter version of sentence1)
sentence8 = ['eats', 'fish'] # Invalid: No NP at sentence start (syntax issue)

print(f"--- Parsing {sentence1} ---")
parser_agreement = AgreementAwareParser(grammar, 'S', terminal_features) # New instance for each parse
tree1 = parser_agreement.parse(sentence1)
if tree1:
    print("Parse successful with agreement. Parse Tree:")
    print(tree1.pretty_print())
else:
    print("Parse failed (syntax or agreement issue).")

print(f"\n--- Parsing {sentence2} ---")
parser_agreement = AgreementAwareParser(grammar, 'S', terminal_features) # New instance for each parse
tree2 = parser_agreement.parse(sentence2)
if tree2:
    print("Parse successful with agreement. Parse Tree:")
    print(tree2.pretty_print())
else:
    print("Parse failed (syntax or agreement issue).")

print(f"\n--- Parsing {sentence3} ---")
parser_agreement = AgreementAwareParser(grammar, 'S', terminal_features) # New instance for each parse
tree3 = parser_agreement.parse(sentence3)
if tree3:
    print("Parse successful with agreement. Parse Tree:")
    print(tree3.pretty_print())
else:
    print("Parse failed (syntax or agreement issue).")

print(f"\n--- Parsing {sentence4} ---")
parser_agreement = AgreementAwareParser(grammar, 'S', terminal_features) # New instance for each parse
tree4 = parser_agreement.parse(sentence4)
if tree4:
    print("Parse successful with agreement. Parse Tree:")
    print(tree4.pretty_print())
else:
    print("Parse failed (syntax or agreement issue).")

print(f"\n--- Parsing {sentence5} ---")
parser_agreement = AgreementAwareParser(grammar, 'S', terminal_features) # New instance for each parse
tree5 = parser_agreement.parse(sentence5)
if tree5:
    print("Parse successful with agreement. Parse Tree:")
    print(tree5.pretty_print())
else:
    print("Parse failed (syntax or agreement issue).")

print(f"\n--- Parsing {sentence6} ---")
parser_agreement = AgreementAwareParser(grammar, 'S', terminal_features) # New instance for each parse
tree6 = parser_agreement.parse(sentence6)
if tree6:
    print("Parse successful with agreement. Parse Tree:")
    print(tree6.pretty_print())
else:
    print("Parse failed (syntax or agreement issue).")

print(f"\n--- Parsing {sentence7} ---")
parser_agreement = AgreementAwareParser(grammar, 'S', terminal_features) # New instance for each parse
tree7 = parser_agreement.parse(sentence7)
if tree7:
    print("Parse successful with agreement. Parse Tree:")
    print(tree7.pretty_print())
else:
    print("Parse failed (syntax or agreement issue).")

print(f"\n--- Parsing {sentence8} ---")
parser_agreement = AgreementAwareParser(grammar, 'S', terminal_features) # New instance for each parse
tree8 = parser_agreement.parse(sentence8)
if tree8:
    print("Parse successful with agreement. Parse Tree:")
    print(tree8.pretty_print())
else:
    print("Parse failed (syntax or agreement issue).")
```

EXP 18-Implement a simple FOPC parser for basic logical expressions using python program.

```python
import re

# Simple FOPC parser
def parse_fopc(expression):
    print("Input Expression:", expression)
    print("\nParsed Components:")

    # Find quantifiers
    quantifiers = re.findall(r'\b(forall|exists)\b', expression)

    # Find predicates
    predicates = re.findall(r'([A-Za-z]+)\s*\(([^)]*)\)', expression)

    # Find logical operators
    operators = re.findall(r'(AND|OR|NOT|IMPLIES)', expression)

    # Display quantifiers
    if quantifiers:
        print("Quantifiers:", quantifiers)
    else:
        print("Quantifiers: None")

    # Display predicates
    if predicates:
        print("Predicates:")
        for name, arguments in predicates:
            print(" ", name, "->", arguments.split(","))
    else:
        print("Predicates: None")

    # Display operators
    if operators:
        print("Logical Operators:", operators)
    else:
        print("Logical Operators: None")


# Example FOPC expression
expression = "forall x (Human(x) IMPLIES Mortal(x))"

parse_fopc(expression)
```

EXP 19-Create a program for word sense disambiguation using the Lesk algorithm using python

```python
import nltk
from nltk.wsd import lesk
from nltk.corpus import wordnet

# Download required WordNet data
nltk.download('wordnet')
nltk.download('omw-1.4')

# Given sentence
sentence = "I went to the bank to deposit money"

# Convert sentence into words
words = sentence.split()

# Perform Word Sense Disambiguation
sense = lesk(words, "bank")

# Display result
print("Sentence:", sentence)
print("Target Word: bank")

if sense:
    print("\nSelected Sense:", sense.name())
    print("Definition:", sense.definition())
    print("Examples:", sense.examples())
else:
    print("No sense found.")
```

EXP 20-Implement a basic information retrieval system using TF-IDF (Term Frequency-Inverse
Document Frequency) for document ranking using python.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Collection of documents
documents = [
    "Python is a popular programming language",
    "Python is used for machine learning",
    "Machine learning is a branch of artificial intelligence",
    "Artificial intelligence is used in many applications"
]

# User query
query = "Python machine learning"

# Create TF-IDF vectorizer
vectorizer = TfidfVectorizer()

# Convert documents into TF-IDF vectors
tfidf_matrix = vectorizer.fit_transform(documents)

# Convert query into TF-IDF vector
query_vector = vectorizer.transform([query])

# Calculate similarity between query and documents
similarity = cosine_similarity(query_vector, tfidf_matrix).flatten()

# Rank documents based on similarity
ranking = similarity.argsort()[::-1]

# Display results
print("Query:", query)
print("\nDocument Ranking:")

for rank, index in enumerate(ranking, start=1):
    print(f"{rank}. Document {index + 1}")
    print(f"   Similarity Score: {similarity[index]:.4f}")
    print(f"   Text: {documents[index]}")
```
